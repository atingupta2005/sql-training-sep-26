# Session 1 - Relational Model and Basic SQL

Welcome. In this session we will learn:

- how company data is stored in **tables**
- how tables **link** to each other
- how to ask simple questions with **SQL**

Sample tables are in the database **`training`** on **your** MySQL. Use names like `training.customers`.


---
# Chapter 1 - The story of our data


Imagine a company that sells IT products to other companies.

Every day the team asks questions like:

- Which customers live in South India?
- Which Dell laptops are low on stock?
- How many orders are still pending?

Those answers live in tables inside database `training`.


**What lives in the `training` database**

| Area | What it holds |
|------|----------------|
| **Catalog** | Products and categories, prices and stock |
| **Customers** | Company accounts, city / country / region |
| **Orders and money** | Orders, line items, payments, staff who sold |


### Count rows in the main tables


````sql
SELECT
  (SELECT COUNT(*) FROM training.customers) AS customers,
  (SELECT COUNT(*) FROM training.products) AS products,
  (SELECT COUNT(*) FROM training.orders) AS orders,
  (SELECT COUNT(*) FROM training.payments) AS payments;
````

---
# Chapter 2 - What is a table?


Think of a table like a clean Excel sheet with rules.

- Each **column** has a fixed meaning
- Each **row** is one item (one customer, one order, ...)


**A table = named columns + rows of facts**

Example: `training.customers`

| customer_id | company_name | city | country |
|------------:|--------------|------|---------|
| 1 | Nimbus Softwares Pvt Ltd | Mumbai | India |
| 2 | Helios FinServe Pvt Ltd | Bengaluru | India |
| 3 | Apex Technologies Pvt Ltd | Hyderabad | India |


### Browse customers


````sql
SELECT customer_id, company_name, city, country, region
FROM training.customers
ORDER BY customer_id
LIMIT 10;
````

---
# Chapter 3 - How tables link together


A customer can place many orders. We do **not** copy the full company details into every order.

Instead, `orders.customer_id` points to `customers.customer_id`.


**Primary key and foreign key**

| Table | Column | Role |
|-------|--------|------|
| `customers` | `customer_id` | **primary key** |
| `orders` | `order_id` | **primary key** |
| `orders` | `customer_id` | **foreign key** -> `customers.customer_id` |

One customer can have many orders.


**One-to-many relationships**

- One **customer** -> many **orders**
- One **order** -> many **order_details**
- One **product** -> many **order_details**
- One **category** -> many **products**


### Browse orders with a customer id


````sql
SELECT order_id, customer_id, employee_id, order_date, status, freight
FROM training.orders
WHERE customer_id IS NOT NULL
ORDER BY order_id
LIMIT 8;
````

# Chapter 4 - Database and your workspace


In **MySQL** on your laptop:

1. You connect to your MySQL **server**
2. Inside it you find **databases**
3. Inside a database you find **tables**

For this course, sample data is in database **`training`**.

```text
MySQL (your laptop)
└── database: training
    ├── customers
    ├── products
    ├── orders
    ├── order_details
    ├── payments
    └── ... other sample tables
```

Write tables like this: `training.customers`.


**How tables connect (one to many)**

| From | To | Relationship |
|------|----|--------------|
| departments | employees | one to many |
| employees | orders | one to many |
| customers | orders | one to many |
| orders | order_details | one to many |
| products | order_details | one to many |
| categories | products | one to many |
| orders | payments | one to many |


### Product categories


````sql
SELECT category_id, category_name, description
FROM training.categories
ORDER BY category_id;
````

### Browse products


````sql
SELECT product_id, product_name, unit_price, units_in_stock, discontinued
FROM training.products
ORDER BY product_id
LIMIT 10;
````

**Order status values in the sample data**

```text
pending -> processing -> shipped -> delivered
                         \-> cancelled
```


### How many orders are in each status?


````sql
SELECT status, COUNT(*) AS n
FROM training.orders
GROUP BY status
ORDER BY n DESC;
````

### Spot messy data counts


````sql
SELECT
  (SELECT COUNT(*) FROM training.customers WHERE email IS NULL) AS missing_emails,
  (SELECT COUNT(*) FROM training.orders WHERE customer_id IS NULL) AS orphan_orders,
  (SELECT COUNT(*) FROM training.products WHERE discontinued = 1) AS discontinued_skus;
````

### Customers with no email on file


````sql
SELECT customer_id, company_name, city, country, email
FROM training.customers
WHERE email IS NULL
ORDER BY customer_id
LIMIT 20;
````

# Chapter 5 - Asking questions with SELECT


**How to read a query (step by step)**

```text
FROM   -> pick the table
WHERE  -> keep matching rows
SELECT -> keep chosen columns
ORDER  -> sort
LIMIT  -> first N rows
```


### Customers in India


````sql
SELECT customer_id, company_name, city, region, country
FROM training.customers
WHERE country = 'India'
ORDER BY city, company_name
LIMIT 15;
````

### Customers in Bengaluru or Hyderabad


````sql
SELECT customer_id, company_name, city, country
FROM training.customers
WHERE city IN ('Bengaluru', 'Hyderabad', 'Chennai')
ORDER BY city, company_name
LIMIT 20;
````

---
# Chapter 6 - AND and OR


**AND vs OR**

| Operator | Meaning | Result size |
|----------|---------|-------------|
| **AND** | both must be true | smaller |
| **OR** | either can be true | larger |


### Customers in India (West or South)


````sql
SELECT customer_id, company_name, city, region
FROM training.customers
WHERE country = 'India'
  AND (region = 'West' OR region = 'South')
ORDER BY region, city
LIMIT 20;
````

### Expensive products with low stock


````sql
SELECT product_id, product_name, unit_price, units_in_stock
FROM training.products
WHERE unit_price >= 50000
  AND units_in_stock < 25
  AND discontinued = 0
ORDER BY units_in_stock, unit_price DESC
LIMIT 15;
````

### Shipped orders in March 2023


````sql
SELECT order_id, customer_id, order_date, status, freight
FROM training.orders
WHERE order_date BETWEEN '2023-03-01' AND '2023-03-31'
  AND status = 'shipped'
ORDER BY order_date, order_id
LIMIT 20;
````

# Chapter 7 - LIKE patterns (text search)


**LIKE patterns**

| Pattern | Meaning |
|---------|---------|
| `'Dell%'` | starts with Dell |
| `'%Technologies%'` | contains Technologies |
| `'%Ltd'` | ends with Ltd |


### Products that start with Dell


````sql
SELECT product_id, product_name, unit_price
FROM training.products
WHERE product_name LIKE 'Dell%'
ORDER BY product_name
LIMIT 15;
````

### Companies with Technologies in the name


````sql
SELECT customer_id, company_name, city, country
FROM training.customers
WHERE company_name LIKE '%Technologies%'
ORDER BY company_name
LIMIT 15;
````

### High freight open orders


````sql
SELECT order_id, order_date, status, freight
FROM training.orders
WHERE status <> 'cancelled'
  AND freight >= 350
ORDER BY freight DESC
LIMIT 15;
````

### List distinct countries


````sql
SELECT DISTINCT country
FROM training.customers
ORDER BY country;
````

# Chapter 8 - NULL and missing values


**NULL means missing or unknown**

| Wrong | Correct |
|-------|---------|
| `WHERE email = NULL` | `WHERE email IS NULL` |


### Friendly label for missing email


````sql
SELECT customer_id,
       company_name,
       COALESCE(email, '(no email on file)') AS email_display
FROM training.customers
WHERE customer_id BETWEEN 1 AND 20
ORDER BY customer_id;
````

### Orders in June 2023


````sql
SELECT order_id, order_date, status,
       EXTRACT(YEAR FROM order_date) AS order_year,
       EXTRACT(MONTH FROM order_date) AS order_month
FROM training.orders
WHERE EXTRACT(YEAR FROM order_date) = 2023
  AND EXTRACT(MONTH FROM order_date) = 6
ORDER BY order_date
LIMIT 15;
````

### Orders not shipped yet


````sql
SELECT order_id, customer_id, order_date, status, shipped_date
FROM training.orders
WHERE shipped_date IS NULL
  AND status IN ('pending', 'processing')
ORDER BY order_date
LIMIT 15;
````

# Chapter 9 - CASE labels


**CASE turns numbers into labels**

| Band | Rule |
|------|------|
| budget | `unit_price < 5000` |
| mid | 5000 to 24999 |
| premium | `unit_price >= 25000` |


### Price bands with CASE


````sql
SELECT product_id, product_name, unit_price,
       CASE
         WHEN unit_price < 5000 THEN 'budget'
         WHEN unit_price < 25000 THEN 'mid'
         ELSE 'premium'
       END AS price_band
FROM training.products
WHERE discontinued = 0
ORDER BY unit_price DESC
LIMIT 20;
````
